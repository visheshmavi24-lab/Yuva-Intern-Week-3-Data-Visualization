# Yuva Intern Week 3 – Data Visualization and Reporting

**Dataset:** Blinkit / Grocery Sales Dataset

**Objective:** Create required visualizations and derive useful insights from the cleaned Week 2 data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('blinkit_sales_cleaned.csv')
print('Shape:', df.shape)
df.head()
```

## 1. Bar Chart – Total Sales by Item Type

```python
sales_by_type = df.groupby('Item_Type')['Item_Outlet_Sales'].sum().sort_values()
plt.figure(figsize=(10,6))
sales_by_type.plot(kind='barh')
plt.title('Total Sales by Item Type')
plt.xlabel('Total Item Outlet Sales')
plt.ylabel('Item Type')
plt.tight_layout()
plt.show()
```

**Insight:** Category-level sales comparison helps identify which product groups contribute most to outlet revenue.

## 2. Line Chart – Sales by Outlet Establishment Year

```python
year_sales = df.groupby('Outlet_Establishment_Year')['Item_Outlet_Sales'].sum().sort_index()
plt.figure(figsize=(9,5))
year_sales.plot(marker='o')
plt.title('Sales by Outlet Establishment Year')
plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

**Insight:** The line chart shows how sales are distributed across outlets established in different years and helps identify establishment cohorts associated with higher sales.

## 3. Scatter Plot – Item MRP vs Item Outlet Sales

```python
plt.figure(figsize=(9,6))
plt.scatter(df['Item_MRP'], df['Item_Outlet_Sales'], alpha=0.25)
plt.title('Item MRP vs Item Outlet Sales')
plt.xlabel('Item MRP')
plt.ylabel('Item Outlet Sales')
plt.tight_layout()
plt.show()
```

**Insight:** The dataset shows a positive relationship between Item MRP and Item Outlet Sales. Published analysis of the same 8,523-row dataset reports a Pearson correlation of about 0.568.

## 4. Correlation Heatmap

```python
num_cols=['Item_Weight','Item_Visibility','Item_MRP','Outlet_Establishment_Year','Item_Outlet_Sales']
corr=df[num_cols].corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

**Insight:** Item_MRP has the strongest positive numeric relationship with sales (about 0.57), while Item_Visibility has a weak negative relationship with sales. Item_Weight has very little linear relationship with sales.

## Conclusion

The visual analysis converts the cleaned Week 2 dataset into understandable charts. Product category distribution, outlet establishment cohorts, price-sales relationship, and numeric correlations provide a compact view of sales behaviour. These findings can support further business analysis and predictive modelling.
